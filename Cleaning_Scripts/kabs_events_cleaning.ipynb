```python
#adjusting the right data types for the columns

# 1. Dates
date_cols = [
    "Date of Report",
    "Date of Start of Outbreak/Event",
    "Date Event Reported"
]

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors="coerce")


# 2. Numeric/count variables
numeric_cols = [
    "Number Affected",
    "Number Affected.1",
    "Number Affected.2",
    "Number Affected.3",
    "Number Affected.4",
    "Number Affected.5",
    "Number Affected.6",
    "Number Affected.7",
    "Number Affected.8",
    "Number Affected.9",
    "Number Affected.10",
    "Number at Risk",
    "Number Sick / Bitten",
    "Number Dead",
    "Number of Humans Affected (If zoonosis)",
    "Number Slaughtered",
    "Number Destroyed",
    "Number Vaccinated",
    "Number Sick",
    "Longitude",
    "Latitude"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")


# 3. Categorical variables
categorical_cols = [
    "Source Form Version",
    "Submitter Organization",
    "Submitter Role",
    "County",
    "Sub-County",
    "Ward",
    "Locality",
    "Species Affected",
    "Other Species",
    "Production System",
    "Abortion",
    "Sudden Death",
    "Death",
    "Hemorrhagic Signs",
    "Neurologic signs",
    "Animal Bites",
    "Respiratory Signs",
    "Oral/Foot Lesions",
    "Cutaneous/Skin Lesions",
    "Gastrointestinal tract syndromes",
    "Other Syndromes",
    "Disease/Condition",
    "Other Disease",
    "Nature of Diagnosis",
    "Test Used",
    "Disease Control Method",
    "Other Control Method"
]

for col in categorical_cols:
    df[col] = df[col].astype("category")


# 4. Identifiers
id_cols = [
    "Submitter UUID",
    "Submitter Username",
    "Submitter Name",
    "Id"
]

for col in id_cols:
    df[col] = df[col].astype("string")
```

```python
#checking the null values in the dataset
null_values = df.isnull().sum()

null_summary = pd.DataFrame({
    "Null Count": null_values,
    "Null Percentage": (null_values / len(df) * 100).round(2)
})

null_summary.sort_values(by="Null Percentage", ascending=False)
```

```python
#columns to drop from the dataset

cols_to_drop = [
    "Date Event Reported",
    "Test Used",
    "Number Affected.10",
    "Number Affected.4",
    "Number Affected.1",
    "Number Affected.3",
    "Number Affected",
    "Number Affected.2",
    "Number Affected.6",
    "Number Affected.7",
    "Number Affected.8",
    "Number Affected.9",
    "Number Affected.5",
    "Death",
    "Location",
    "Location.1",
    "Submitter UUID",
    "Submitter Username",
    "Submitter Name",
    "Id",
    "Source Form Version",
    "Submitter Organization",
    "Submitter Role",
    "Hemorrhagic Signs",
    "Neurologic signs",
    "Animal Bites",
    "Respiratory Signs",
    "Oral/Foot Lesions",
    "Cutaneous/Skin Lesions",
    "Gastrointestinal tract syndromes",
    "Other Syndromes",
    "Number Destroyed",
    "Number Vaccinated",
    "Date of Start of Outbreak/Event",
    "Production System",
    "Number Slaughtered",
    "Abortion",
    "Sudden Death"
]

df1 = df.drop(columns=cols_to_drop)

print(f"\nRemaining dataset shape: {df1.shape}")
```

```python
#fill 0 in the numericals variables

num_cols = ["Number Sick / Bitten", "Number Sick", "Number of Humans Affected (If zoonosis)"]
df1[num_cols] = df1[num_cols].fillna(0)
```

```python
#filling the nulls for coordinates

# Fill hierarchy: Ward -> Sub-County -> County
for group_col in ['Ward', 'Sub-County', 'County']:
    df1['Latitude'] = df1['Latitude'].fillna(
        df1.groupby(group_col)['Latitude'].transform('mean')
    )
    df1['Longitude'] = df1['Longitude'].fillna(
        df1.groupby(group_col)['Longitude'].transform('mean')
    )

# Check how many are still missing after all three levels
print(df1[['Latitude', 'Longitude']].isna().sum())
```

```python
#fill the nulls in nature of diagnosis as "unkown"

df1['Nature of Diagnosis'] = df1['Nature of Diagnosis'].cat.add_categories('Unknown')
df1['Nature of Diagnosis'] = df1['Nature of Diagnosis'].fillna('Unknown')
```

```python
df1.isnull().sum().sort_values()
```

```python
#mapping the "other" columns to the main variables

mapping = {
    'Disease/Condition': 'Other Disease',
    'Disease Control Method': 'Other Control Method',
    'Species Affected': 'Other Species'
}

for main_col, other_col in mapping.items():
    # drop categorical dtype temporarily so assignment isn't restricted
    df1[main_col] = df1[main_col].astype(object)
    df1[other_col] = df1[other_col].astype(object)

    mask = df1[main_col].isna() | (df1[main_col] == 'Other')
    df1.loc[mask, main_col] = df1.loc[mask, other_col]
```

```python
# dropping the other columns

df2 = df1.drop(columns=['Other Disease', 'Other Control Method', 'Other Species'])
```

```python
#fill unkown values to  Disease/condition
if isinstance(df1['Disease/Condition'].dtype, pd.CategoricalDtype):
    df2['Disease/Condition'] = df2['Disease/Condition'].cat.add_categories('Unknown')

df2['Disease/Condition'] = df2['Disease/Condition'].fillna('Unknown')
```
