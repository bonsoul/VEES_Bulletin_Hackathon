```python
columns_to_drop = [
    "Submitter UUID",
    "Other sign",
    "Other sign.1",
    "Other Sign",
    "Other Species",
    "Specify Other Sign",
    "Specify Other Sign.1",
    "Submitter Username",
    "Submitter Name",
    "Submitter Organization",
    "Submitter Role",
    "Id",
    "Date of Start of sickness",
    "Location",
    "Location.1"
]

comm_df1 = comm_df.drop(columns=columns_to_drop)
```

```python
# set the correct data types

comm_df1["Date of Report"] = pd.to_datetime(comm_df1["Date of Report"], errors="coerce")
comm_df1["Longitude"] = pd.to_numeric(comm_df1["Longitude"], errors="coerce").astype("float64")
comm_df1["Latitude"] = pd.to_numeric(comm_df1["Latitude"], errors="coerce").astype("float64")

count_columns = [
    "Total Number of Animals in the herd",
    "Number Sick",
    "Number Dead"
]
comm_df1[count_columns] = comm_df1[count_columns].apply(
    pd.to_numeric, errors="coerce"
).astype("Int64")

text_columns = comm_df1.select_dtypes(include="object").columns
comm_df1[text_columns] = comm_df1[text_columns].astype("string")

comm_df1.dtypes
```
