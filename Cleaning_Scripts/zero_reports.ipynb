```python
# Import pandas for loading and cleaning the dataset.
import pandas as pd
```

```python
# Load the zero-report dataset.
zero_report = pd.read_excel(
    r"C:\Users\ADMIN\Documents\VEES_Bulletin_Hackathon\Raw_Datasets\Zero Report_KABS.xls"
)
```

```python
# Count missing values in each column.
zero_report.isnull().sum().sort_values(ascending=False)
```

```python
# List the available columns before selecting fields to remove.
zero_report.columns
```

```python
# Remove metadata and unused fields from the zero-report dataset.
columns_to_drop = [
    "Submitter UUID",
    "Submitter Username",
    "Submitter Name",
    "Submitter Organization",
    "Submitter Role",
    "Id",
    "Date of Report",
    "Location.1",
    "Location",
    "Other Species",
    "Notes",
]

zero_clean_df = zero_report.drop(columns=columns_to_drop, errors="ignore")
zero_clean_df.shape
```

```python
# Fill numeric columns with 0 and text or categorical columns with "Unknown".
numeric_columns = zero_clean_df.select_dtypes(include="number").columns
zero_clean_df[numeric_columns] = zero_clean_df[numeric_columns].fillna(0)

categorical_columns = zero_clean_df.select_dtypes(include="category").columns
for column in categorical_columns:
    if "Unknown" not in zero_clean_df[column].cat.categories:
        zero_clean_df[column] = zero_clean_df[column].cat.add_categories(["Unknown"])
    zero_clean_df[column] = zero_clean_df[column].fillna("Unknown")

string_columns = zero_clean_df.select_dtypes(include=["string", "object"]).columns
zero_clean_df[string_columns] = zero_clean_df[string_columns].fillna("Unknown")

# Confirm that no missing values remain.
zero_clean_df.isnull().sum().sort_values(ascending=False)
```

```python
# Save the cleaned zero-report dataset as a CSV file.
zero_clean_df.to_csv("../Cleaned Data/zero_cleaned.csv", index=False)
```
