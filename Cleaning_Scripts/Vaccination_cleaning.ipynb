```python
# import libraries
import numpy as np
import pandas as pd
```

```python
#load the data
df = pd.read_excel(r"C:\Users\ADMIN\Documents\VEES_Bulletin_Hackathon\Datasets\reports_2026-08-18.xls")
```

```python
# drop the inappropriate columns
df1 = df.drop(columns=["Id","Latitude.1", "Longitude.1","Submitter Name", "Submitter Role", "Submitter Username", "Submitter UUID"], errors="ignore")
```

```python
# numeric columns: keep as NaN (proper missing value)
num_cols = ["Latitude", "Longitude"]
df1[num_cols] = df1[num_cols].fillna(np.nan)

# string columns: fill with "Unknown"
str_cols = ["Specify other disease", "Vaccination Site(s)", "Submitter Organization"]
df1[str_cols] = df[str_cols].fillna("Unknown")
```

```python
# save the cleaned data into vaccination cleaned
df1.to_csv(r"C:\Users\ADMIN\Documents\VEES_Bulletin_Hackathon\Cleaned Data\vaccination_cleaned.csv", index=False)
```
