```python
# Remove unused metadata, rescue fields, and detailed affected-count columns.
columns_to_drop = [
    "Location.1",
    "Date of rescue",
    "Other Area Type",
    "Bloody Diarrhoea",
    "Post Mortem(PM) done",
    "Evidence of bleeding",
    "Site Species rescued from",
    "Site Species rescued to",
    "Number of species rescued",
    "Date of Report",
    "Location",
    "Test Used",
    "Unnamed: 52",
    "Submitter Organization",
    "Unnamed: 68",
    "Submitter Role",
    "Submitter Username",
    "Submitter UUID",
    "Submitter Name",
    "Species rescued",
    "Number Affected.11",
    "Number Affected.10",
    "Number Affected.9",
    "Number Affected.9",
    "Number Affected.8",
    "Number Affected.7",
    "Number Affected.6",
    "Number Affected.5",
    "Number Affected.4",
    "Number Affected.3",
    "Number Affected.2",
    "Number Affected.1",
    "PM findings",
    "Other Control Method",
    "Other Species",
]

# Ignore missing names so the cleaning cell remains reusable.
wild_clean_df = wild_df.drop(columns=columns_to_drop, errors="ignore")
wild_clean_df.shape
```

```python
wild_clean_df.isnull().sum().sort_values(ascending=False)
```

```python
# Convert date fields to datetime for date-based analysis.
date_columns = [
    "Date of Start of Outbreak/Event",
    "Date Event Reported",
]

# Convert coordinates and event counts to numeric values.
numeric_columns = [
    "Latitude",
    "Longitude",
    "Number Affected",
    "Number Sick / Bitten",
    "Number Dead",
    "Number of Humans Affected",
]

for column in date_columns:
    if column in wild_clean_df.columns:
        wild_clean_df[column] = pd.to_datetime(
            wild_clean_df[column], errors="coerce"
        )

for column in numeric_columns:
    if column in wild_clean_df.columns:
        wild_clean_df[column] = pd.to_numeric(
            wild_clean_df[column], errors="coerce"
        )

# Store remaining text fields as pandas string dtype.
text_columns = wild_clean_df.select_dtypes(include=["str"]).columns
wild_clean_df[text_columns] = wild_clean_df[text_columns].astype("string")

wild_clean_df.dtypes
```

```python
# Fill numeric fields, including coordinates, with zero.
numeric_columns = wild_clean_df.select_dtypes(include="number").columns
wild_clean_df[numeric_columns] = wild_clean_df[numeric_columns].fillna(0)

# Add and use "Unknown" for missing categorical values.
categorical_columns = wild_clean_df.select_dtypes(include="category").columns
for column in categorical_columns:
    if "Unknown" not in wild_clean_df[column].cat.categories:
        wild_clean_df[column] = wild_clean_df[column].cat.add_categories(["Unknown"])
    wild_clean_df[column] = wild_clean_df[column].fillna("Unknown")

# Fill missing text values with a readable placeholder.
string_columns = wild_clean_df.select_dtypes(include=["string", "object"]).columns
wild_clean_df[string_columns] = wild_clean_df[string_columns].fillna("Unknown")
```

```python
# Save the cleaned wildlife dataset.
wild_clean_df.to_csv("../Cleaned Data/wildlife_cleaned_df.csv", index=False)
```
