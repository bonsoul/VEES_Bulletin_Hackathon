```python
#import libraries

import pandas as pd
import numpy as np
```

```python
from pathlib import Path

# Locate the project folder whether the notebook is run from the repo root or Cleaning_Scripts.
project_root = Path.cwd().resolve()
if project_root.name == "Cleaning_Scripts":
    project_root = project_root.parent

data_path = project_root / "Hackathon_Data" / "monthly_data.xls"
df = pd.read_excel(data_path)

# Keep only the required columns.
df_clean = df[[
    "Disease/Condition",
    "County",
    "Number Sick",
    "Number Dead",
]].copy()

output_path = project_root / "Hackathon_Data" / "monthly_cleaned.csv"
df_clean.to_csv(output_path, index=False)

print(f"Saved cleaned data to: {output_path}")
print(f"Dataset shape: {df_clean.shape}")
df_clean.head()
```
