```python
from pathlib import Path
import pandas as pd

# Read data
project_root = Path.cwd().resolve()

if project_root.name == "Cleaning_Scripts":
    project_root = project_root.parent

excel_candidates = [
    project_root / "Hackathon_Data" / "weeklydata.xls",
    project_root / "Hackathon_Data" / "weeklydata.xlsx",
]

excel_path = next(
    (path for path in excel_candidates if path.exists()),
    None
)

if excel_path is None:
    raise FileNotFoundError(
        f"Weekly data not found in {project_root / 'Hackathon_Data'}"
    )

df1 = pd.read_excel(excel_path)

# Preview the data
print(f"File loaded: {excel_path}")
print(f"Dataset shape: {df1.shape}")
print(df1.head())
```

```python
#dataset columns

df1.columns
```

```python
# Keep only the required columns
df_clean = df1[[
    "Disease/Condition",
    "County",
    "Number Sick",
    "Number Dead"
]].copy()

# Rename them to shorter names
df_clean = df_clean.rename(columns={
    "Disease/Condition": "Disease/Condition",
    "Number Sick": "No. Sick",
    "Number Dead": "No. Dead"
})

# Save the cleaned data in the project data folder
cleaned_path = project_root / "Hackathon_Data" / "weekly_cleaned.csv"
df_clean.to_csv(cleaned_path, index=False)

df_clean.head()
```
