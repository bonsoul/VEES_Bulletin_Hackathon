```python

# Replace "Other" with the corresponding free-text entry
data.loc[data["Disease/Condition"].eq("Other"), "Disease/Condition"] = data["Other Disease"]

mask = data["Disease Control Method"].str.contains("Other", na=False)
data.loc[mask, "Disease Control Method"] = (
    data.loc[mask].apply(
        lambda row: row["Disease Control Method"].replace("Other", row["Other Control Method"]),
        axis=1
    )
)

# Combine into final columns and remove the "Other" columns
data.drop(columns=["Other Disease", "Other Control Method"], inplace=True)
```

```python
data2 = data.drop(columns=[c for c in data.columns if c.startswith("Number Affected") or c.startswith("Location")])
```

```python
species_map = {
    # Rabbits
    "rabbit": "Rabbits",
    "rabbits": "Rabbits",
    "rabit": "Rabbits",
    "rabbit kitten": "Rabbits",

    # Poultry
    "chicken": "Chicken",
    "broiler": "Chicken",
    "broilers": "Chicken",
    "poultry": "Chicken",
    "avian": "Chicken",
    "duck": "Ducks",
    "ducks": "Ducks",
    "turkey": "Turkeys",

    # Dogs
    "dog": "Dogs",
    "dogs": "Dogs",
    "canine": "Dogs",
    "puppies": "Dogs",

    # Goats
    "goat": "Goats",
    "goats": "Goats",
    "shoat": "Goats",
    "shoats": "Goats",
    "kid": "Goats",
    "kids": "Goats",
    "goat lambs": "Goats",
    "sheep and goat": "Goats",

    # Sheep
    "lamb": "Sheep",
    "lambs": "Sheep",

    # Cattle
    "calf": "Cattle",

    # Other
    "camel": "Camel",
    "horse": "Horse",
    "fish": "Fish",
    "insecta": "Insect",
    "donkey": "Donkey",
    "zebra": "Zebra",
    "eland": "Eland",
}

# Clean and harmonise Other Species
data2["Other Species"] = (
    data2["Other Species"]
    .astype("string")
    .str.strip()
    .str.lower()
)

data2["Other Species"] = data2["Other Species"].replace(species_map)

# Capitalize anything not already mapped, and tidy spacing
data2["Other Species"] = (
    data2["Other Species"]
    .str.strip()
    .str.title()
)

# Check harmonised results
data2["Other Species"].value_counts()
```
